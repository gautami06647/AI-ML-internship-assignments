```python
import pandas as pd

data = {
    "review": [
        "I loved this movie, the acting was amazing and story was great",
        "This movie was terrible, boring plot and poor acting",
        "An excellent film with strong performances",
        "Worst movie I have ever watched",
        "The movie was good but could have been better",
        "Amazing direction and beautiful cinematography",
        "Not worth watching at all",
        "The story was engaging and emotional",
        "Bad movie with weak storyline",
        "Fantastic movie, I enjoyed every moment"
    ] * 200   # creates 2000 rows
}

df = pd.DataFrame(data)
print("Total rows in dataset:", len(df))
df.head()
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

df["tokens"] = df["review"].apply(word_tokenize)
df.head()
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

df["stemmed_words"] = df["tokens"].apply(
    lambda words: [ps.stem(w) for w in words]
)

df.head()
```

```python
for i in range(3):
    print("Original Review:")
    print(df["review"][i])
    print("\nTokens:")
    print(df["tokens"][i])
    print("\nStemmed Words:")
    print(df["stemmed_words"][i])
    print("-" * 60)
```
